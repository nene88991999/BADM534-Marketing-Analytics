```python
#%pip install requests
import requests

# the review to be analyzed, which can also be read from a file or other source
review = "I was drawn to these ALLSWIFIT sneakers because they're super cute and look very similar to some of the more expensive brands. I ordered them in two colors because I loved the style so much! The pink/orange combo is really vibrant and fun. My initial impression is that they're very lightweight and breathable, which is a plus. However, I did notice that they're a little bit loose on my feet. I'm a true size 10, and while these fit length-wise, there's some extra room. I'm hesitant to size down, though, because I worry that a smaller size might be too tight around the ankle and footbed, especially since some other reviewers mentioned the opening being snug. I'll probably add some non-slip heel pads to see if that helps with the looseness. Despite the slight sizing issue, I'm still happy with them so far. They're comfortable enough for casual wear, and I'm hoping the heel pads will solve the minor fit problem. For the price, they're a great dupe for the more expensive styles."


# Gemini endpoint: adding the api_key you obtained from your google AI studio account
url = "https://generativelanguage.googleapis.com/v1beta/models/gemini-2.0-flash:generateContent?key=YOUR_API_KEY"

# Prompt instructing Gemini to classify the review
# The prompt can be changed to suit your analysis needs

prompt = f"""
Given the following product review, label it with:
1. Comfort level: High, Medium, or Low
2. Weight: Light, Medium, or Heavy
3. Appearance: Stylish, Vibrant, Plain, or Dull

Review:
\"\"\"{review}\"\"\"

Respond with just the three labels, in this format:
Comfort: <label>, Weight: <label>, Appearance: <label>
"""

# Format the request body
headers = {"Content-Type": "application/json"}
data = {
    "contents": [
        {"parts": [{"text": prompt}]}
    ]
}

# Send the request
response = requests.post(url, headers=headers, json=data)
result = response.json()

# Extract and print the result
try:
    output = result['candidates'][0]['content']['parts'][0]['text']
    print("Gemini Classification:", output)
except Exception as e:
    print("Error in response:", result)
```

```python
import pandas as pd

# Parse the output string to extract the labels
labels = dict(item.strip().split(": ") for item in output.strip().split(", "))

# Create a dataframe with the review and its labels
df = pd.DataFrame([{
    "Review": review,
    "Comfort": labels.get("Comfort"),
    "Weight": labels.get("Weight"),
    "Appearance": labels.get("Appearance")
}])

# Save the dataframe to a CSV file
df.to_csv("review_classification.csv", index=False)
```
