```python
import pandas as pd
```

```python
# read the CSV file into a DataFrame df
df = pd.read_csv('CoffeeData.csv')
```

```python
# obtain the basic summary statistics of df
summary = df.describe()
print(summary)
```

```python
# calculate the mean for each column in the df
mean_values = df.mean()
print(mean_values)

print("Variance for each column in the df\n", df.var())
print("Median value for each column in the df\n", df.median())
```
