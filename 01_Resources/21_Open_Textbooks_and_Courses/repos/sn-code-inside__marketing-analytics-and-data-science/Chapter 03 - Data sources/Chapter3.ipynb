```python
# !pip install tweepy 
import pandas as pd
import tweepy
# Replace with your actual Bearer Token from X Developer Portal
bearer_token = "AAAAAAAAAAAAAAAAAAAAAMED"

client = tweepy.Client(bearer_token=bearer_token)

# Example: Search 100 recent tweets about electric vehicles
query = "electric vehicles"
response = client.search_recent_tweets(
    query=query,
    max_results=100,
    tweet_fields=["created_at", "author_id", "text", "lang"]
)

# # Print results
# for tweet in response.data:
#     print(f"[{tweet.created_at}] ({tweet.lang}) {tweet.text}\n")
if response.data:
    # Convert to a list of dicts
    tweets = [{
        "created_at": tweet.created_at,
        "author_id": tweet.author_id,
        "lang": tweet.lang,
        "text": tweet.text
    } for tweet in response.data]

    # Create a DataFrame
    df = pd.DataFrame(tweets)

    # Save to CSV
    df.to_csv("electric_vehicles_tweets.csv", index=False, encoding="utf-8-sig")

    print("Tweets saved to 'electric_vehicles_tweets.csv'")
else:
    print("No tweets found for your query.")
```
