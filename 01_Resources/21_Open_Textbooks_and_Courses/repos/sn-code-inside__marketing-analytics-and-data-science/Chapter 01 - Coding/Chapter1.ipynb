## Read in datafile: CoffeeData.csv
In order for the code to be able to find the data file easily, we recommend users to put them in the same file folder. The python code will look for a file automatically in the same file folder.

```python
# First, install the required packages:
# pip install pandas
# Then, import the necessary library
import pandas as pd
```

```python
# read the CSV file into a DataFrame df
df = pd.read_csv('CoffeeData.csv')
```

```python
# show the first 10 rows of the DataFrame
df.head(2)
```

## Access colums in the data

We can access the columns in the data by calling the name of the variable, using quotation marks

```python
df.columns
```

```python
# calculate the correlation between price and sales
correlation = df['Price'].corr(df['Sales'])
print(f"Correlation between Price and Sales: {correlation}")
```

## Create some simple plots to visualize the data

In order to make plots, we need to import another library, called matplotlib

```python
import matplotlib.pyplot as plt
```

### Line Plots

```python
# plot the sales variable using line plot, using red color

plt.plot(df['Sales'], color='red')
plt.title('Sales Over Time')
```

```python
# Make two line plots together, for sales and price, in the same figure
plt.plot(df['Price'], color='blue', label='Price')
plt.plot(df['Sales'], color='red', label='Sales')
plt.xlabel('Week')

# Add a legend to the plot
plt.legend()
# Show the plot 
plt.show()
```

Analyze the plot: 
- It barely shows the variations in the price variable, as the scale is too much smaller comparing to the sales variable. 
- Solution: rescale the sales variable before plotting them together

To do that, we need to first create a new variable of sales, in a different unit, in thousands

```python
df['Salesk'] = df['Sales'] / 1000
plt.plot(df['Price'], color='blue', label='Price')
plt.plot(df['Salesk'], color='red', label='Sales')
plt.xlabel('Week')

# Add a legend to the plot
plt.legend()
# Show the plot 
plt.show()
```

### Scatter plots

```python
# Create a scatter plot of price vs sales
plt.scatter(df['Price'], df['Sales'], color='green')
plt.title('Price vs Sales')
plt.xlabel('Price')
plt.ylabel('Sales')
# Show the plot
plt.show()
```

### Histograms

```python
# create a histogram of the sales variable, and the price variable
plt.hist(df['Sales'], bins=20, color='purple', alpha=0.7)
plt.title('Sales Histogram')    
plt.xlabel('Sales')
plt.ylabel('Frequency')
# Show the plot
plt.show()
plt.hist(df['Price'], bins=20, color='orange', alpha=0.7)
plt.title('Price Histogram')
plt.xlabel('Price')
plt.ylabel('Frequency')
# Show the plot
plt.show()
```

```python
## Basic Statistics
# Calculate the mean, median, and standard deviation of the sales variable
mean_sales = df['Sales'].mean()
median_sales = df['Sales'].median()
std_sales = df['Sales'].std()
print(f"Mean Sales: {mean_sales}")
print(f"Median Sales: {median_sales}")
print(f"Standard Deviation of Sales: {std_sales}")
```

```python
# sort the DataFrame by sales in descending order
df_sorted = df.sort_values(by='Sales', ascending=False)
df_sorted.head()
```
