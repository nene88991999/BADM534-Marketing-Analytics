```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)

# Simulate samples
n_samples = 100000
beta1 = np.random.beta(5, 7, n_samples)
beta2 = np.random.beta(4, 8, n_samples)
beta3 = np.random.beta(3, 9, n_samples)

# Find which distribution is largest for each sample
winners = np.argmax(np.vstack([beta1, beta2, beta3]), axis=0)

# Compute probabilities
prob_beta1 = np.mean(winners == 0)
prob_beta2 = np.mean(winners == 1)
prob_beta3 = np.mean(winners == 2)

print(f"Probability beta(5,7) is largest: {prob_beta1:.4f}")
print(f"Probability beta(4,8) is largest: {prob_beta2:.4f}")
print(f"Probability beta(3,9) is largest: {prob_beta3:.4f}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
plt.scatter(beta1, beta2, alpha=0.3, s=1)
plt.xlabel('beta1 (Beta(5,7))')
plt.ylabel('beta3 (Beta(3,9))')
plt.title('Scatter plot of beta1 vs beta3')
plt.plot([0, 1], [0, 1], color='black', linestyle='--', linewidth=2, label='45° line')
plt.show()
```

```python
beta4 = np.random.beta(50, 70, n_samples)
beta5 = np.random.beta(30, 90, n_samples)
plt.figure(figsize=(8, 6))
plt.scatter(beta4, beta5, alpha=0.3, s=1)
plt.xlabel('beta4 (Beta(50,70))')
plt.ylabel('beta5 (Beta(30,90))')
plt.title('Scatter plot of beta1 vs beta2')
plt.plot([0, 1], [0, 1], color='black', linestyle='--', linewidth=2, label='45° line')
plt.show()
```
